```python
import sys

class SovereignLCompiler:
    def __init__(self):
        # Establish the 3 physical machine binary control tokens (Section 4.1)
        self.TOKENS = {
            'CONNECT':  0x01,  # Establishes hardwired causal memory tracks
            'RESONATE': 0x02,  # Triggers the 8-neighbor integer division split
            'COLLAPSE': 0x03   # Global energy conservation power kill-switch
        }

        # Define the valid 4D integer matrix boundaries for orthogonal tracking
        self.valid_axes = {'x', 'y', 'z', 'w'}

    def tokenize_line(self, line_num, line_text):
        """Scans raw incoming text code strings into binary hardware tokens."""
        # Clean the text stream of comments and empty spacing
        cleaned = line_text.split('#')[0].strip()
        if not cleaned:
            return None

        # Split into command and argument tokens
        parts = cleaned.split()
        command = parts[0].upper()

        if command not in self.TOKENS:
            print(f"COMPILER ERROR [Line {line_num}]: Invalid instruction signature '{command}'")
            return None

        binary_token = self.TOKENS[command]
        arguments = parts[1:]

        # Validate arguments according to structural hardware claims
        if command == 'CONNECT':
            # Expects coordinate shifts like: CONNECT x+1 y-1
            if len(arguments) < 2:
                print(f"COMPILER ERROR [Line {line_num}]: CONNECT requires at least 2 target nodes.")
                return None
            print(f"LEXER REGISTERED: Token {hex(binary_token)} (CONNECT) -> Target Nodes: {arguments}")
            return (binary_token, arguments)

        elif command == 'RESONATE':
            # Expects injection target and value: RESONATE (0,0,0,0) 27
            if len(arguments) != 2:
                print(f"COMPILER ERROR [Line {line_num}]: RESONATE requires (target_node) and [pulse_value].")
                return None
            try:
                pulse_val = int(arguments[1])
                print(f"LEXER REGISTERED: Token {hex(binary_token)} (RESONATE) -> Target: {arguments[0]} | Value: {pulse_val}")
                return (binary_token, arguments[0], pulse_val)
            except ValueError:
                print(f"COMPILER ERROR [Line {line_num}]: Pulse value must be a whole-number integer.")
                return None

        elif command == 'COLLAPSE':
            # Expects target system tracking loop: COLLAPSE global_net
            if len(arguments) != 1:
                print(f"COMPILER ERROR [Line {line_num}]: COLLAPSE requires a target network domain tracking token.")
                return None
            print(f"LEXER REGISTERED: Token {hex(binary_token)} (COLLAPSE) -> Scope: {arguments[0]}")
            return (binary_token, arguments[0])

    def compile_source(self, source_code):
        """Processes full multi-line Sovereign-L source scripts into bytecode streams."""
        print("=== SOVEREIGN-L COMPILER INITIALIZATION SEQUENCE ===")
        bytecode_stream = []
        lines = source_code.strip().split('\n')

        for idx, line in enumerate(lines, 1):
            compiled_token = self.tokenize_line(idx, line)
            if compiled_token:
                bytecode_stream.append(compiled_token)

        print(f"=== COMPILATION COMPLETE: Generated {len(bytecode_stream)} Hardware Bytecode Nodes ===\n")
        return bytecode_stream

# --- Execute Sovereign-L Script Compilation Test ---
compiler_unit = SovereignLCompiler()

# Sample Sovereign-L instruction code text representing your 4D matrix routing rules
sovereign_script = """
# Foundational Code Script for UFNI Spacetime Routing Logic
CONNECT x+1 y-1 z+0 w+0      # Wire physical path to orthogonal neighbors
RESONATE (0,0,0,0) 27        # Fire integer split array across 8 data buses
COLLAPSE global_matrix       # Trip kill-switch if energy conservation drops
"""

# Compile the string code into raw machine arrays
compiled_bytecode = compiler_unit.compile_source(sovereign_script)
```
